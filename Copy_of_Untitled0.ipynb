```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Arun", "Bala", "Charan", "Deepak"],
    "Marks": [85, 70, 95, 80]
}

df = pd.DataFrame(data)

print(df)

plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.savefig("chart.png")

df.to_csv("cleaned_data.csv", index=False)

print("Completed")
```
